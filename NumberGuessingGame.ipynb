# Goal: Build a Number guessing game


````python
# Number Guessing Game
# Created: Tiago Fernandes
# Last updated: 04NOV25

# === IMPORTS ===
import random

# === CODE ===

import random

while True:
    number = random.randint(1, 10)

    # Try-except for initial guess with range validation
    while True:
        try:
            guess = int(input("Guess a number between 1 and 10: "))
            if 1 <= guess <= 10:
                break
            else:
                print("Please enter a number between 1 and 10!")
        except ValueError:
            print("Invalid input! Please enter a number.")

    lives = 3

    while guess != number and lives > 0:
        if guess < number:
            print("Too low!")
            lives -= 1
            print("Lives remaining:", lives)
        elif guess > number:
            print("Too high!")
            lives -= 1
            print("Lives remaining:", lives)

        if lives > 0 and guess != number:
            # Try-except for subsequent guesses with range validation
            while True:
                try:
                    guess = int(input("Guess again: "))
                    if 1 <= guess <= 10:
                        break
                    else:
                        print("Please enter a number between 1 and 10!")
                except ValueError:
                    print("Invalid input! Please enter a number.")

        if guess == number:
            print("Congratulations! You guessed the number", number)
            break

    if lives == 0:
        print("You ran out of lives. The number was", number)

    # Simple restart logic
    start_again = input("Would you like to start again? (y/n): ").upper()

    if start_again == "Y":
        continue
    elif start_again == "N":
        print("Thanks for playing, see you next time!")
        break
    else:
        print("Incorrect response, see you next time!")
        break
````

# Number Guessing Game - Learning Summary

## What We Built
A number guessing game where:
- The computer picks a random number (1-10)
- Player gets 3 lives to guess correctly
- Game gives "too high" or "too low" hints
- Player can restart after winning or losing
- Invalid inputs are handled gracefully
- Out-of-range numbers are rejected

---

## Challenges We Faced & Solutions

### 1. Syntax Errors
**Problem:** `While` instead of `while`, missing quotes, wrong operators (`=` vs `==`)

**Lesson:** Python is case-sensitive and strict about syntax. Always use lowercase for keywords like `while`, `if`, `elif`

### 2. Logic Flow Issues
**Problem:** The "start again" prompt appeared after every wrong guess instead of at the end

**Solution:** Moved it outside the inner `while` loop by unindenting

**Lesson:** **Indentation controls scope** in Python. Code inside a loop repeats; code outside runs after the loop ends

### 3. Duplicate Input Requests
**Problem:** Asked for guess twice when adding try-except

**Solution:** Removed the first `input()` line, kept only the try-except version

**Lesson:** When refactoring code, remove old versions to avoid duplication

### 4. `.upper()` Placement
**Problem:** Applied `.upper()` to the prompt string instead of the user's response

**Solution:** `input("...").upper()` not `input("...".upper())`

**Lesson:** Method chaining - `.upper()` applies to whatever is returned by `input()`, not the string inside it

### 5. Restart Logic Confusion
**Problem:** Initial validation loop checked for valid input but didn't handle what to do with each response correctly

**Solution:** Simplified to use `if/elif/else` with appropriate `continue` and `break` statements:
- "Y" → `continue` (restart game)
- "N" → `break` with goodbye message
- Anything else → `break` with error message

**Lesson:** `continue` vs `break` in nested loops:
- `continue` restarts the **current loop** (outer game loop = new game)
- `break` exits the **current loop** (outer game loop = end program)
- Don't over-complicate validation when simple is better

### 6. Game Continuing After Lives Run Out
**Problem:** Game kept asking for guesses even when lives reached 0

**Solution:** Added condition `if lives > 0 and guess != number:` before asking for next guess

**Lesson:** **Check ALL relevant conditions** before requesting input:
- Don't just check if lives > 0
- Also check if the player hasn't already won (guess != number)
- Multiple conditions with `and` ensure code only runs when ALL conditions are true

### 7. Out-of-Range Input Validation
**Problem:** Players could enter valid numbers outside the 1-10 range (e.g., 100, -5)

**Solution:** Added range validation after try-except:
```python
try:
    guess = int(input("Guess a number between 1 and 10: "))
    if 1 <= guess <= 10:
        break
    else:
        print("Please enter a number between 1 and 10!")
except ValueError:
    print("Invalid input! Please enter a number.")
```

**Lesson:** **Input validation has multiple layers:**
1. **Type validation** (try-except) - Is it the right data type?
2. **Range validation** (if statement) - Is it within acceptable bounds?
3. Order matters: validate type first, then range (can't check range of a string!)

## Key Takeaways for Next Time

### 1. Loop Structure Matters
- Outer `while True:` = game restart loop
- Inner `while guess != number and lives > 0:` = guessing loop
- **Watch your indentation!**

### 2. Input Validation (Multi-Layer Approach)
- **Layer 1:** Use try-except when converting strings to numbers with `int()`
  - Catches ValueError when user enters text
- **Layer 2:** Validate the range/bounds of the input
  - Use `if 1 <= value <= 10:` for range checking
- **Layer 3:** Consider if validation loops are necessary or if simple if/else is clearer
- **Always validate in order:** Type → Range → Logic

### 3. Testing Strategy
- Test edge cases (what happens at lives = 1? lives = 0?)
- Test invalid inputs (letters instead of numbers)
- Test boundary values (0, 1, 10, 11, negative numbers)
- Test the full game flow (win, lose, restart, quit)
- Test different responses to "start again" (y, n, random text)
- **Play through the entire game multiple times** to catch logic bugs

### 4. Code Organization
```python
while True:              # Outer loop - restart game
    setup_game()
    
    while playing:       # Inner loop - game rounds
        play_round()
    
    end_game_logic()     # Outside inner loop, inside outer loop
    
    # Restart decision
    if user_wants_restart:
        continue         # Goes back to start of outer loop
    else:
        break           # Exits outer loop, ends program
```

### 5. Control Flow Keywords
- `break` - exits the current loop immediately
- `continue` - skips rest of current iteration, goes to next iteration
- Use these strategically to control game flow

### 6. Compound Conditions
- Use `and` to combine multiple conditions: `if lives > 0 and guess != number:`
- ALL conditions must be true for the code block to execute
- Think about edge cases: what if lives = 0? what if already guessed correctly?

### 7. Validation Best Practices
```python
# Good validation structure:
while True:
    try:
        value = int(input("Enter number: "))  # Type conversion
        if min_value <= value <= max_value:    # Range check
            break                               # Valid input, exit loop
        else:
            print("Out of range!")              # Range error message
    except ValueError:
        print("Invalid type!")                  # Type error message
```
